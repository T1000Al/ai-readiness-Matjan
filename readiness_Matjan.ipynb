```python
def calculate_stats(numbers):
    mean = sum(numbers) / len(numbers)
    maximum = max(numbers)
    return mean, maximum
```

```python
numbers = [10, 20, 30, 40, 50]
```

```python
mean, maximum = calculate_stats(numbers)
```

```python
print("Mean:", mean)
print("Maximum:", maximum)
```
